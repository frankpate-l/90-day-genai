```python
### Accessing List Elements

fruits=["apple","banana","cherry","kiwi","gauva"]
```

```python
fruits[0]
```

```python
print(fruits[1])
print(fruits[-1])
print(fruits[3])
print(fruits[-3])
print(fruits[4])
```

```python
print(fruits[1 : 3])
print(fruits[-1 : -3])
print(fruits[2:3])
print(fruits[1:])
```

```python
#modifying list elements
print(fruits)
fruits[1]="watermelon"
print(fruits)

```
