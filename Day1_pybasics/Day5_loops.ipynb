```python
# while loop

# The while loop continues to execute as long as the condition is True.

count=0

while count<5:
    print(count)
    count=count+1
```

```python
#loop contorl statements 
#break


for i in range(10):
    if i==7:
        break
    print(i)
```

```python
#continue
#The continue statement skips the current iteration and continues with the next.

for i in range(20):
    if i%2==0:
        continue
    print(i)
    
```

```python
#pass 
# it does nothing

for i in range(5):
    if i==3:
        print("num is",i )
        pass
    print(i)
```

```python
# nested loops

for i in range(4):
    for j in range(3):
        print(f"i:{i} and j:{j}")
```

```python
## Examples- Calculate the sum of first N natural numbers using a while and for loop

## while loop 

n=10
sum=0
count=1

while count<=n:
    sum=sum+count
    count=count+1
print(f"sum of 10 num is {sum}")
```

```python
#for loop
sum=0
n=11
for i in range(n):
    sum=sum+i

print(f"sum 10 num is {sum}")
```

```python
## Example- Prime numbers between 1 and 100

for i in range(1,101):
    if i>1:
        for j in range(2,i):
            if i%j==0:
                break
        else:
            print(i)
                
            
```
